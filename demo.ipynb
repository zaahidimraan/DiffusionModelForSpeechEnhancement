# Resample Audio

```python
import os
import librosa
import soundfile as sf

def resample_audio_folder(folder_path, target_sample_rate=48000):
    """Resamples all audio files in a folder to the target sample rate.

    Args:
        folder_path (str): Path to the folder containing audio files.
        target_sample_rate (int, optional): The desired sample rate. Defaults to 48000 Hz.
    """
    for filename in os.listdir(folder_path):
        if filename.endswith((".wav", ".mp3", ".ogg", ".flac")):  # Add other supported formats if needed
            file_path = os.path.join(folder_path, filename)
            
            # Load audio with original sample rate
            y, sr = librosa.load(file_path, sr=None)  # Load with original sample rate
            
            # Check if resampling is necessary
            if sr != target_sample_rate:
                print(f"Resampling {filename} from {sr} Hz to {target_sample_rate} Hz...")
                y_resampled = librosa.resample(y, orig_sr=sr, target_sr=target_sample_rate)

                # Overwrite the original file with the resampled audio
                sf.write(file_path, y_resampled, target_sample_rate)  
            else:
                print(f"Skipping {filename} (already at {target_sample_rate} Hz)")

# Get folder path from user input
folder_path = input("Enter the path to the folder containing audio files: ")

# Resample audio files
resample_audio_folder(folder_path)

print("Resampling complete!")
```
